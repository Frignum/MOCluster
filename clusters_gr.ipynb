```python
import pandas as pd
import plotly.graph_objects as go


month_cols = [
    c for c in df.columns
    if c.startswith(('23-', '24-'))
]

clusters_long = df.melt(
    id_vars=['region_name', 'mo'],
    value_vars=month_cols,
    var_name='month',
    value_name='cluster'
)

clusters_long = clusters_long.dropna(subset=['cluster']).copy()
clusters_long['cluster'] = clusters_long['cluster'].astype(int)



df_plot = df_2.copy()

# если в df_2 есть year и month:
df_plot['month'] = (
    (df_plot['year'] % 100).astype(str).str.zfill(2)
    + '-'
    + df_plot['month'].astype(str).str.zfill(2)
)


df_plot = clusters_long.merge(
    df_plot,
    on=['region_name', 'mo', 'month'],
    how='inner'
)

df_plot = df_plot.dropna(
    subset=[
        'Средняя зарплата',
        'Продовольствие_share'
    ]
)



cluster_colors = {
    0: '#E53935',
    1: '#FDD835',
    2: '#43A047',
    3: '#1E88E5',
    4: '#FB8C00',
    5: '#8E44AD'
}


```

```python
profile_jan_2023 = (
    df_plot[df_plot['month'] == '23-01']
    .dropna(subset=['cluster'])
    .groupby('cluster')[
        [
            'Средняя зарплата',
            'population',
            'urban_share',
            'share_industry',
            'share_agriculture',
            'share_services',
            'Здоровье_share',
            'Маркетплейсы_share',
            'Общественное питание_share',
            'Продовольствие_share',
            'Транспорт_share'
        ]
    ]
    .mean()
    .round(3)
)

profile_jan_2023
```

```python
cluster_mapping = {
    3: 0,
    1: 1,
    4: 2,
    2: 3,
    0: 4
}

df_plot['cluster'] = df_plot['cluster'].replace(cluster_mapping)
```

```python
cluster_share = (
    df_plot
    .dropna(subset=['cluster'])
    .groupby(['month', 'cluster'])
    .size()
    .groupby(level=0)
    .transform(lambda x: x / x.sum() * 100)
    .reset_index(name='share')
)

fig = go.Figure()


# Явно задаём порядок кластеров
for cluster in sorted(cluster_share['cluster'].unique()):
    d = cluster_share[cluster_share['cluster'] == cluster]

    fig.add_trace(go.Scatter(
        x=d['month'],
        y=d['share'],
        mode='lines',
        name=f'Кластер {cluster}',
        stackgroup='one',
        line=dict(color=cluster_colors[cluster]),
        legendrank=-cluster          # ← ключевой параметр
    ))

fig.update_layout(
    title='Изменение структуры кластеров',
    xaxis_title='Месяц',
    yaxis_title='Доля муниципалитетов, %',
    yaxis=dict(range=[0, 100]),
    template='plotly_white',
    autosize=True,          # ← вместо height=550
    margin=dict(l=40, r=20, t=50, b=40)  # чтобы не было лишних полей
)

fig.show()
```

```python
fig.write_html('cluster_profiles_change_2023.html')
```

```python
df_plot1 = df_plot[df_plot['Средняя зарплата'] <= 200000]
fig = go.Figure()

clusters = sorted(df_plot1['cluster'].unique())

for month in month_cols:

    month_df = df_plot1[
        df_plot1['month'] == month
    ]

    for cluster in clusters:

        d = month_df[
            month_df['cluster'] == cluster
        ]

        fig.add_trace(
            go.Scatter(
                x=d['Средняя зарплата'],
                y=d['Продовольствие_share'],

                mode='markers',

                name=f'Кластер {cluster}',

                marker=dict(
                    color=cluster_colors[cluster],
                    size=8,
                    opacity=0.75
                ),

                text=d.apply(
                    lambda r:
                    f"{r['region_name']}<br>"
                    f"{r['mo']}<br>"
                    f"Кластер: {r['cluster']}<br>"
                    f"Зарплата: {r['Средняя зарплата']:,.0f} руб.<br>"
                    f"Продовольствие: {r['Продовольствие_share']:.3f}",
                    axis=1
                ),

                hovertemplate='%{text}<extra></extra>',

                visible=(month == month_cols[0])
            )
        )


# ============================================================
# 6. Ползунок
# ============================================================

n_clusters = len(clusters)

steps = []

for i, month in enumerate(month_cols):

    visible = [False] * len(fig.data)

    start = i * n_clusters
    end = start + n_clusters

    visible[start:end] = [True] * n_clusters

    steps.append(
        dict(
            method='update',

            args=[
                {'visible': visible},

                {
                    'title':
                    f'Распределение муниципальных образований — {month}'
                }
            ],

            label=month
        )
    )


fig.update_layout(

    title=f'Распределение муниципальных образований — {month_cols[0]}',

    xaxis=dict(
        title='Средняя зарплата, руб.',
        tickformat=',',
        range=[20000, 200000]      # ← фиксированная шкала X
    ),

    yaxis=dict(
        title='Продовольствие_share',
        range=[0.2, 0.7]            # ← фиксированная шкала Y
    ),

    sliders=[
        dict(
            active=0,
            currentvalue={
                'prefix': 'Месяц: ',
                'font': {'color': '#2f5d3a', 'size': 14}
            },
            bgcolor='#dff5b5',
            bordercolor='#2f5d3a',
            borderwidth=1,
            tickcolor='#2f5d3a',
            font=dict(color='#2f5d3a', size=11),
            activebgcolor='#a3e635',
            pad={'t': 30},
            steps=steps
        )
    ],

    legend=dict(title='Кластер'),

    template='plotly_white',

    autosize=True,
)
```

```python
fig.write_html(
    'wage_scatter_cluster.html',
    include_plotlyjs=True
)
```

```python
import plotly.graph_objects as go
import pandas as pd


y_features = [
    'Продовольствие_share',
    'Маркетплейсы_share',
    'Общественное питание_share',
    'Здоровье_share',
    'Транспорт_share',
    'urban_share',
    'share_industry',
    'share_agriculture',
    'share_services',
    'population',
]

labels = {
    'Средняя зарплата': 'Средняя зарплата, руб.',
    'population': 'Население, чел.',
    'urban_share': 'Доля городского населения',
    'share_industry': 'Доля занятых в промышленности',
    'share_agriculture': 'Доля занятых в сельском хозяйстве',
    'share_services': 'Доля занятых в сфере услуг',
    'Здоровье_share': 'Доля расходов на здоровье',
    'Маркетплейсы_share': 'Доля расходов на маркетплейсы',
    'Общественное питание_share': 'Доля расходов на общественное питание',
    'Продовольствие_share': 'Доля расходов на продовольствие',
    'Транспорт_share': 'Доля расходов на транспорт',
}

x_feat = 'Средняя зарплата'
y0 = 'Продовольствие_share'

for col in ['Средняя зарплата', 'population']:
    if col in df_plot.columns:
        df_plot[col] = df_plot[col].round(0).astype('int32')

share_cols = [
    'Здоровье_share', 'Маркетплейсы_share', 'Общественное питание_share',
    'Продовольствие_share', 'Транспорт_share',
    'urban_share', 'share_industry', 'share_agriculture', 'share_services',
]
for col in share_cols:
    if col in df_plot.columns:
        df_plot[col] = df_plot[col].round(3).astype('float32')

clusters = sorted(df_plot['cluster'].unique())
month_cols = sorted(df_plot['month'].unique())

# 1.месяц × кластер

fig = go.Figure()

for month in month_cols:
    month_df = df_plot[df_plot['month'] == month]

    for cluster in clusters:
        d = month_df[month_df['cluster'] == cluster]

        fig.add_trace(
            go.Scatter(
                x=d[x_feat],
                y=d[y0],
                mode='markers',
                name=f'Кластер {cluster}',
                legendgroup=f'cluster-{cluster}',
                showlegend=(month == month_cols[0]),
                marker=dict(
                    color=cluster_colors[cluster],
                    size=8,
                    opacity=0.75,
                ),
                # customdata: регион, МО, месяц, кластер
                customdata=d[['region_name', 'mo', 'month', 'cluster']].values,
                # шаблон: %{x} — зарплата, %{y} — текущая Y-переменная
                hovertemplate=(
                    '%{customdata[0]}<br>'
                    '%{customdata[1]}<br>'
                    'Месяц: %{customdata[2]}<br>'
                    'Кластер: %{customdata[3]}<br>'
                    'Зарплата: %{x:,.0f} руб.<br>'
                    'Значение: %{y:,.3f}'
                    '<extra></extra>'
                ),
                visible=(month == month_cols[0]),
            )
        )

n_clusters = len(clusters)
n_traces = len(fig.data)

# 2. Слайдер по месяцам

steps = []
for i, month in enumerate(month_cols):
    visible = [False] * n_traces
    start = i * n_clusters
    end = start + n_clusters
    visible[start:end] = [True] * n_clusters

    steps.append(dict(
        method='update',
        args=[
            {'visible': visible},
            {'title.text': f'Зависимость переменных от средней зарплаты — {month}'},
        ],
        label=month,
    ))

# 3. Кнопки для оси Y

y_cache = {}
for y_feat in y_features:
    y_cache[y_feat] = []
    for month in month_cols:
        month_df = df_plot[df_plot['month'] == month]
        for cluster in clusters:
            d = month_df[month_df['cluster'] == cluster]
            y_cache[y_feat].append(d[y_feat].tolist())

y_buttons = []
for y_feat in y_features:
    y_buttons.append(dict(
        label=labels[y_feat],
        method='update',
        args=[
            {'y': y_cache[y_feat]},                     # только y
            {'yaxis.title.text': labels[y_feat]},       # только текст оси
        ],
    ))

# ============================================================
# 4. Layout
# ============================================================

fig.update_layout(
    title=dict(
        text=f'Зависимость переменных от средней зарплаты — {month_cols[0]}',
        x=0.5,
        xanchor='center',
        y=0.97,
    ),
    template='plotly_white',
    autosize=True,

    margin=dict(t=140, l=60, r=30, b=100),

    xaxis=dict(title=labels[x_feat], tickformat=','),
    yaxis=dict(title=labels[y0]),

    legend=dict(title='Кластер'),

    updatemenus=[
        dict(
            buttons=y_buttons,
            direction='down',
            showactive=True,
            x=0.0, xanchor='left',
            y=1.14, yanchor='top',
            bgcolor='#dff5b5',
            bordercolor='#2f5d3a',
            font=dict(color='#2f5d3a', size=12),
            pad={'r': 10, 't': 5},
        ),
    ],

    sliders=[
        dict(
            active=0,
            currentvalue={
                'prefix': 'Месяц: ',
                'font': {'color': '#2f5d3a', 'size': 14},
            },
            bgcolor='#dff5b5',
            bordercolor='#2f5d3a',
            borderwidth=1,
            tickcolor='#2f5d3a',
            font=dict(color='#2f5d3a', size=11),
            activebgcolor='#a3e635',
            pad={'t': 40},
            steps=steps,
        )
    ],
)

# ============================================================
# 5. Экспорт
# ============================================================

fig.write_html(
    'wage_scatter_cluster.html',
    include_plotlyjs='cdn',
    full_html=True,
    config={
        'responsive': True,
        'displayModeBar': False,
    },
)
```
